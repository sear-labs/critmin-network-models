```python
import os
params = {
"WLSACCESSID": os.environ.get("GUROBI_WLSACCESSID"),
"WLSSECRET": os.environ.get("GUROBI_WLSSECRET"),
"LICENSEID": int(os.environ.get("GUROBI_LICENSEID")),
}

env = gp.Env(params=params)
```
